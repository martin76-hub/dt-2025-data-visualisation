```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Students - Leht1.csv")

avg_hours = df.groupby("Grade")["Hours of Study"].mean().reindex(["A", "B", "C"])

avg_hours.plot(kind="bar", color="#4a90e2", figsize=(6, 4))

plt.title("Average Hours of Study by Grade", fontsize=14, fontweight="bold")
plt.xlabel("Grade")
plt.ylabel("Average Hours of Study")
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.savefig("hours_vs_grade_chart.png", dpi=300)
plt.show()
```

```python
hobby_counts = df["Hobby"].value_counts().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
hobby_counts.plot(kind="bar", color="#00bfa5")

plt.title("Number of Students by Hobby", fontsize=14, fontweight="bold")
plt.xlabel("Hobby")
plt.ylabel("Number of Students")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.savefig("students_by_hobby_chart.png", dpi=300)
plt.show()
```

The chart shows the popularity of every hobby and how many students participate in them, with soccer being the most popular, with 4 students, and reading, video games, basketball and movies having 2 students participate and the rest having one.

```python
grade_counts = df["Grade"].value_counts().reindex(["A", "B", "C"])

plt.figure(figsize=(5, 5))
plt.pie(
    grade_counts,
    labels=grade_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["#4a90e2", "#00bfa5", "#f5a623"]
)
plt.title("Grade Distribution", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("grade_distribution_pie.png", dpi=300)
plt.show()
```

The chart shows the grade distribution in the glass by percentage, with A being the most popular with 40%, B with 35% and C with 25%.

```python
hobby_counts_top = df["Hobby"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    hobby_counts_top,
    labels=hobby_counts_top.index,
    autopct="%1.1f%%",
    startangle=90,
    shadow=True
)
plt.title("Hobby Distribution", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("hobby_distribution_pie.png", dpi=300)
plt.show()
```

The chart shows the popularity of every hobby by percentage, with soccer being the most popular, with 20%, and reading, video games, basketball and movies having 10% participation and the rest having 5%.

```python


grade_map = {"A": 3, "B": 2, "C": 1}
df["Grade Score"] = df["Grade"].map(grade_map)

plt.figure(figsize=(6, 5))
plt.scatter(df["Hours of Study"], df["Grade Score"], color="#4a90e2", s=80, alpha=0.7, edgecolors="black")

plt.title("Scatter Plot: Hours of Study vs. Grade", fontsize=14, fontweight="bold")
plt.xlabel("Hours of Study")
plt.ylabel("Grade Score (A=3, B=2, C=1)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.savefig("scatter_hours_vs_grade.png", dpi=300)
plt.show()
```

The scatter plots chart shows hours of study correspond to the grade a student will achieve for it, with 4 hours or above receiving a definite A, 3 hours receiving an A or B, 2 hours receiving a B or C and anything below 2 hours receiving a C.

```python
plt.figure(figsize=(7, 5))
plt.scatter(
    df["Age"],
    df["Hours of Study"],
    s=df["Grade Score"] * 200,  # Bubble size based on grade
    alpha=0.6,
    c=df["Grade Score"],
    cmap="viridis",
    edgecolors="black"
)

plt.title("Bubble Chart: Age vs. Hours of Study", fontsize=14, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Hours of Study")
plt.colorbar(label="Grade Score (A=3, B=2, C=1)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.savefig("bubble_age_vs_hours.png", dpi=300)
plt.show()
```

The bubble chart shows the age of the students, their study time and how and the grade they receive. The chart shows that the least amount of study time was by students aged 20 and 21. The most study time was also shared by the ages 20 and 21.

```python
plt.figure(figsize=(7, 5))
df.boxplot(column="Hours of Study", by="Grade", patch_artist=True,
           boxprops=dict(facecolor="#4a90e2", color="black"),
           medianprops=dict(color="red", linewidth=2),
           grid=False)

plt.title("Box Plot: Hours of Study by Grade", fontsize=14, fontweight="bold")
plt.suptitle("")  # Removes default pandas title
plt.xlabel("Grade")
plt.ylabel("Hours of Study")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.savefig("boxplot_hours_by_grade.png", dpi=300)
plt.show()
```

The box plot chart compares the hours of study for students with grades A, B, and C. Students with grade A generally study more hours, while those with grade C have the least study time and more variation.

```python
plt.figure(figsize=(7, 5))
df.boxplot(column="Age", by="Grade", patch_artist=True,
           boxprops=dict(facecolor="#00bfa5", color="black"),
           medianprops=dict(color="red", linewidth=2),
           grid=False)

plt.title("Box Plot: Age Distribution by Grade", fontsize=14, fontweight="bold")
plt.suptitle("")
plt.xlabel("Grade")
plt.ylabel("Age")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.savefig("boxplot_age_by_grade.png", dpi=300)
plt.show()
```

The box plot chart compares the ages of students by grade. It shows that students of all grades have similar age ranges, but those earning grade A include both younger and slightly older students.
