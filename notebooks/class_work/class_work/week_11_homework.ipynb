```python
#Task 1

text = "In the beginning God created the heaven and the earth. And the earth was without form, and void; and darkness was upon the face of the deep. And the Spirit of God moved upon the face of the waters. And God said, Let there be light: and there was light. And God saw the light, that it was good: and God divided the light from the darkness. And God called the light Day, and the darkness he called Night. And the evening and the morning were the first day."

# Using the code below I split the text and print the first 15 words.
words = text.split()
print("First 15 words:", words[:15])
print("Total number of words:", len(words))

# By indexing and slicing, I select words or sentence parts that I want to be printed out.
print("The 10th word:", words[9])
print("Every 5th word:", words[::5])
print("Sample sentence:", " ".join(words[20:30]))
```

```python
#Task 2

from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

text = "In the beginning God created the heaven and the earth. And the earth was without form, and void; and darkness was upon the face of the deep. And the Spirit of God moved upon the face of the waters. And God said, Let there be light: and there was light. And God saw the light, that it was good: and God divided the light from the darkness. And God called the light Day, and the darkness he called Night. And the evening and the morning were the first day."

stopwords = set(STOPWORDS)
stopwords.update(["the", "and", "was"])  # optional cleanup

wordcloud = WordCloud(width=800, height=400,
                      background_color="white",
                      stopwords=stopwords).generate(text)

plt.figure(figsize=(10,5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()

#The word cloud highlights the word "God" as the most used word, as "light", this shows that the text is most likely religious in nature, possibly from the bible and most likely about creation
```
